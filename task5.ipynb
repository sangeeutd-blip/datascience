```python
df["Country/Region"]=pd.to_datetime(df["ObservationDate"],errors="coerce")
df.head()
```

```python
plt.figure(figsize=(10,10))
plt.plot(df['Deaths'],df['Country/Region'])
plt.xlabel('deaths')
plt.ylabel("country")
plt.title("dth VS cntry")
plt.show()
```

```python
value1=df["Deaths"].loc[5]
value2=df["Deaths"].iloc[-9]
if value1<value2:
    print("Decreasing Trend")
elif value1>value2:
    print("Increasing Trend")
else:
    print("Stable Trend")
```
