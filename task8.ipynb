```python
tips['sex']=tips['sex'].fillna(tips['sex'].mode())[0]
tips['smoker']=tips['smoker'].fillna(tips['smoker'].mode())[0]
tips['day']=tips['day'].fillna(tips['day'].mode())[0]
tips['time']=tips['time'].fillna(tips['time'].mode())[0]
```

```python
sns.countplot(data=tips, x='smoker')
plt.title("Count Plot")
plt.show()
```

```python
tips['smoker'].value_counts().plot(kind='bar', color='pink')
plt.xlabel('Categories')
plt.ylabel('Frequency')
plt.show()
```

```python
tips['size'].value_counts().plot(kind='pie', autopct='%1.1f%%', startangle=90)
plt.title("Pie Chart")
plt.xlabel("day")
plt.ylabel("time") 
plt.show()
```
